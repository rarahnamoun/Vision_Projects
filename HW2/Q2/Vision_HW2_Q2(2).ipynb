```python
import cv2
import numpy as np
import pandas as pd
import math
import tensorflow as tf
from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split
from skimage.metrics import structural_similarity as ssim
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.feature import canny
import matplotlib.pyplot as plt

# Load FER2013 dataset
df = pd.read_csv("fer2013.csv")

def preprocess_images(df):
    images = []
    for index, row in df.iterrows():
        pixels_string = row['pixels']
        flat_pixels = list(map(int, pixels_string.split()))
        num_pixels = len(flat_pixels)
        width = int(math.sqrt(num_pixels))
        height = num_pixels // width

        if width * height != num_pixels:
            print(f"Skipping row {index}: Pixel data does not fit perfectly into a square image.")
            continue

        pixel_array = np.array(flat_pixels, dtype=np.uint8).reshape((height, width))
        pixel_array = pixel_array / 255.0  # Normalize the image
        images.append(pixel_array)

    return np.array(images)


images = preprocess_images(df)

if images.size == 0:
    raise ValueError("No valid images after preprocessing. Please check the dataset.")

X_train, X_test = train_test_split(images, test_size=0.2, random_state=42)

def add_noise(images, noise_factor=0.05):
    noisy_images = images + noise_factor * np.random.normal(loc=0.0, scale=1.0, size=images.shape)
    noisy_images = np.clip(noisy_images, 0., 1.)
    return noisy_images

# Add noise to the data
X_train_noisy = add_noise(X_train)
X_test_noisy = add_noise(X_test)

# Reshape the images to include the channel dimension
X_train_noisy = X_train_noisy[..., np.newaxis]
X_test_noisy = X_test_noisy[..., np.newaxis]
X_train = X_train[..., np.newaxis]
X_test = X_test[..., np.newaxis]

# Define the U-Net model for denoising
def build_unet_denoising_model(input_shape):
    inputs = layers.Input(shape=input_shape)

    c1 = layers.Conv2D(64, (3, 3), activation='relu', padding='same')(inputs)
    c1 = layers.BatchNormalization()(c1)
    c1 = layers.Conv2D(64, (3, 3), activation='relu', padding='same')(c1)
    c1 = layers.BatchNormalization()(c1)
    p1 = layers.MaxPooling2D((2, 2))(c1)

    c2 = layers.Conv2D(128, (3, 3), activation='relu', padding='same')(p1)
    c2 = layers.BatchNormalization()(c2)
    c2 = layers.Conv2D(128, (3, 3), activation='relu', padding='same')(c2)
    c2 = layers.BatchNormalization()(c2)
    p2 = layers.MaxPooling2D((2, 2))(c2)

    c3 = layers.Conv2D(256, (3, 3), activation='relu', padding='same')(p2)
    c3 = layers.BatchNormalization()(c3)
    c3 = layers.Conv2D(256, (3, 3), activation='relu', padding='same')(c3)
    c3 = layers.BatchNormalization()(c3)
    p3 = layers.MaxPooling2D((2, 2))(c3)

    c4 = layers.Conv2D(512, (3, 3), activation='relu', padding='same')(p3)
    c4 = layers.BatchNormalization()(c4)
    c4 = layers.Conv2D(512, (3, 3), activation='relu', padding='same')(c4)
    c4 = layers.BatchNormalization()(c4)
    p4 = layers.MaxPooling2D((2, 2))(c4)

    b = layers.Conv2D(1024, (3, 3), activation='relu', padding='same')(p4)
    b = layers.BatchNormalization()(b)
    b = layers.Conv2D(1024, (3, 3), activation='relu', padding='same')(b)
    b = layers.BatchNormalization()(b)

    u4 = layers.Conv2DTranspose(512, (3, 3), activation='relu', strides=(2, 2), padding='same')(b)
    u4 = layers.concatenate([u4, c4])
    c5 = layers.Conv2D(512, (3, 3), activation='relu', padding='same')(u4)
    c5 = layers.BatchNormalization()(c5)
    c5 = layers.Conv2D(512, (3, 3), activation='relu', padding='same')(c5)
    c5 = layers.BatchNormalization()(c5)

    u3 = layers.Conv2DTranspose(256, (3, 3), activation='relu', strides=(2, 2), padding='same')(c5)
    u3 = layers.concatenate([u3, c3])
    c6 = layers.Conv2D(256, (3, 3), activation='relu', padding='same')(u3)
    c6 = layers.BatchNormalization()(c6)
    c6 = layers.Conv2D(256, (3, 3), activation='relu', padding='same')(c6)
    c6 = layers.BatchNormalization()(c6)

    u2 = layers.Conv2DTranspose(128, (3, 3), activation='relu', strides=(2, 2), padding='same')(c6)
    u2 = layers.concatenate([u2, c2])
    c7 = layers.Conv2D(128, (3, 3), activation='relu', padding='same')(u2)
    c7 = layers.BatchNormalization()(c7)
    c7 = layers.Conv2D(128, (3, 3), activation='relu', padding='same')(c7)
    c7 = layers.BatchNormalization()(c7)

    u1 = layers.Conv2DTranspose(64, (3, 3), activation='relu', strides=(2, 2), padding='same')(c7)
    u1 = layers.concatenate([u1, c1])
    c8 = layers.Conv2D(64, (3, 3), activation='relu', padding='same')(u1)
    c8 = layers.BatchNormalization()(c8)
    c8 = layers.Conv2D(64, (3, 3), activation='relu', padding='same')(c8)
    c8 = layers.BatchNormalization()(c8)

    outputs = layers.Conv2D(1, (1, 1), activation='sigmoid')(c8)

    model = models.Model(inputs, outputs)
    return model

def calculate_psnr(original, denoised):
    return psnr(original, denoised, data_range=1.0)

def calculate_ssim(original, denoised, win_size=3):
    return ssim(original.squeeze(), denoised.squeeze(), win_size=win_size, data_range=1.0, channel_axis=None)

def calculate_epi(noisy_image, denoised_image):
    edges_noisy = canny((noisy_image * 255).astype('uint8'))
    edges_denoised = canny((denoised_image * 255).astype('uint8'))
    return np.sum(edges_denoised) / max(np.sum(edges_noisy), 1)

def calculate_igs(noisy_image, denoised_image):
    grad_noisy = np.gradient(noisy_image)
    grad_denoised = np.gradient(denoised_image)
    return np.mean([np.abs(g_n - g_d) for g_n, g_d in zip(grad_noisy, grad_denoised)])

input_shape = (X_train.shape[1], X_train.shape[2], 1)
model = build_unet_denoising_model(input_shape)
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['accuracy'])


class MetricsCallback(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        predictions = self.model.predict(X_test_noisy)

        psnr_values = [calculate_psnr(X_test[i], predictions[i]) for i in range(len(X_test))]
        ssim_values = [calculate_ssim(X_test[i], predictions[i]) for i in range(len(X_test))]
        epi_values = [calculate_epi(X_test_noisy[i].squeeze(), predictions[i].squeeze()) for i in range(len(predictions))]
        igs_values = [calculate_igs(X_test_noisy[i].squeeze(), predictions[i].squeeze()) for i in range(len(predictions))]

        print(f"\nEpoch {epoch + 1}:")
        print(f"  Average PSNR: {np.mean(psnr_values):.4f}, Average SSIM: {np.mean(ssim_values):.4f}")
        print(f"  Average EPI: {np.mean(epi_values):.4f}, Average IGS: {np.mean(igs_values):.4f}")


metrics_callback = MetricsCallback()

history = model.fit(
    X_train_noisy,
    X_train,
    validation_data=(X_test_noisy, X_test),
    epochs=10,
    batch_size=32,
    callbacks=[metrics_callback]
)

# Evaluate the model on the test set
predictions = model.predict(X_test_noisy)

# Calculate final metrics
psnr_values_before = [calculate_psnr(X_test[i], X_test_noisy[i]) for i in range(len(X_test))]
psnr_values_after = [calculate_psnr(X_test[i], predictions[i]) for i in range(len(X_test))]

ssim_values_before = [calculate_ssim(X_test[i], X_test_noisy[i]) for i in range(len(X_test))]
ssim_values_after = [calculate_ssim(X_test[i], predictions[i]) for i in range(len(X_test))]

epi_values = [calculate_epi(X_test_noisy[i].squeeze(), predictions[i].squeeze()) for i in range(len(predictions))]
igs_values = [calculate_igs(X_test_noisy[i].squeeze(), predictions[i].squeeze()) for i in range(len(predictions))]

print(f"Final Metrics:")

print(f"  Average PSNR after denoising: {np.mean(psnr_values_after):.4f}")

print(f"  Average SSIM after denoising: {np.mean(ssim_values_after):.4f}")
print(f"  Average EPI: {np.mean(epi_values):.4f}")
print(f"  Average IGS: {np.mean(igs_values):.4f}")

# Visualize a few examples
def plot_denoising_results(original, noisy, denoised, index):
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.title("Original Image")
    plt.imshow(original.squeeze(), cmap='gray')
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.title("Noisy Image")
    plt.imshow(noisy.squeeze(), cmap='gray')
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.title("Denoised Image")
    plt.imshow(denoised.squeeze(), cmap='gray')
    plt.axis('off')

    plt.tight_layout()
    plt.show()

# Show results for a few samples
for i in range(5):  
    plot_denoising_results(X_test[i], X_test_noisy[i], predictions[i], i)

```
