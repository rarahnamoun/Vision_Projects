```python
import cv2
import numpy as np
import pandas as pd
import math
import os
from google.colab.patches import cv2_imshow

# Load the FER2013 dataset
df = pd.read_csv("fer2013.csv")

# Create directories for clean and noisy images
clean_images_dir = "clean_images"
noisy_images_dir = "noisy_images"

# Create the directories if they don't exist
if not os.path.exists(clean_images_dir):
    os.makedirs(clean_images_dir)
if not os.path.exists(noisy_images_dir):
    os.makedirs(noisy_images_dir)

# Function to add noise to images
def add_noise(image, noise_factor=0.1):
    noisy_image = image + noise_factor * np.random.normal(loc=0.0, scale=1.0, size=image.shape)
    noisy_image = np.clip(noisy_image, 0., 1.)
    return noisy_image

# Loop through the rows of the dataframe and process each image
for index, row in df.iterrows():
    pixels_string = row['pixels']
    flat_pixels = list(map(int, pixels_string.split()))
    num_pixels = len(flat_pixels)
    width = int(math.sqrt(num_pixels))
    height = num_pixels // width

    # Ensure the dimensions fit perfectly into a square
    if width * height != num_pixels:
        raise ValueError(f"Pixel data in row {index} does not fit perfectly into a square image. Please adjust dimensions manually.")

    # Convert flat pixel array to 2D image array
    pixel_array = np.array(flat_pixels, dtype=np.uint8).reshape((height, width))

    # Save the clean image
    clean_img_path = os.path.join(clean_images_dir, f"clean_image_{index}.png")
    cv2.imwrite(clean_img_path, pixel_array)

    # Add noise to the clean image
    noisy_image = add_noise(pixel_array / 255.0)  # Normalize to [0, 1] before adding noise
    noisy_image = (noisy_image * 255).astype(np.uint8)  # Denormalize back to [0, 255]

    # Save the noisy image
    noisy_img_path = os.path.join(noisy_images_dir, f"noisy_image_{index}.png")
    cv2.imwrite(noisy_img_path, noisy_image)

    # Optionally, display the images
    cv2_imshow(noisy_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

    # Print the comma-separated pixel values for the current image
    comma_separated = ",".join(map(str, flat_pixels))
    print(f"Comma-separated pixel values for image {index}:", comma_separated)

print("Image generation complete! Clean and noisy images are saved.")
```

```python

class Conv_block(nn.Module):
    def __init__(self, in_channels, out_channels, size=3, padding=1, dropout=0.2, stride=1):
        super().__init__()
        self.convs = nn.Sequential(
            nn.Conv2d(in_channels, out_channels, size, padding=padding),
            nn.LeakyReLU(0.1, inplace=True),
            nn.Dropout(dropout),
            nn.BatchNorm2d(out_channels),

            nn.Conv2d(out_channels, out_channels, size, padding=padding, stride=stride),
            nn.LeakyReLU(0.1, inplace=True),
            nn.Dropout(dropout),
            nn.BatchNorm2d(out_channels),
        )

    def forward(self, x):
        return self.convs(x)


class DeConv_block(nn.Module):
    def __init__(self, in_channels, out_channels, size=3, padding=1, dropout=0.2):
        super().__init__()
        self.deConv = nn.ConvTranspose2d(in_channels, out_channels, size, padding=1, stride=2, output_padding=1)
        self.convs = Conv_block(out_channels * 2, out_channels, size, padding, dropout)

    def forward(self, x, residual):
        x = self.deConv(x)
        x = torch.cat([x, residual], 1)
        return self.convs(x)


class Unet(nn.Module):
    def __init__(self, d=64, out_channels=3, dropout=0.2):
        super().__init__()

        # Unet encoder
        self.conv_blocks = nn.ModuleList([
            Conv_block(3, d, dropout=dropout),
            Conv_block(d, 2 * d, stride=2, dropout=dropout),
            Conv_block(2 * d, 4 * d, stride=2, dropout=dropout),
            Conv_block(4 * d, 8 * d, stride=2, dropout=dropout),
            Conv_block(8 * d, 16 * d, stride=2, dropout=dropout)
        ])

        # Unet decoder
        self.deConv_blocks = nn.ModuleList([
            DeConv_block(16 * d, 8 * d, dropout=dropout),
            DeConv_block(8 * d, 4 * d, dropout=dropout),
            DeConv_block(4 * d, 2 * d, dropout=dropout),
            DeConv_block(2 * d, d, dropout=dropout),
        ])

        self.output_conv = nn.Conv2d(d, out_channels, 3, padding=1)

    def forward(self, x):
        x1 = self.conv_blocks[0](x)
        x2 = self.conv_blocks[1](x1)
        x3 = self.conv_blocks[2](x2)
        x4 = self.conv_blocks[3](x3)
        x5 = self.conv_blocks[4](x4)

        x = self.deConv_blocks[0](x5, x4)
        x = self.deConv_blocks[1](x, x3)
        x = self.deConv_blocks[2](x, x2)
        x = self.deConv_blocks[3](x, x1)

        x = self.output_conv(x)
        return x


# Define Discriminator
class Discriminator(nn.Module):
    def __init__(self, label_channels=3, input_channels=3, d=64):
        super().__init__()

        self.convs = nn.Sequential(
            nn.Conv2d(label_channels + input_channels, d, 3, 2, 1),
            nn.LeakyReLU(0.1, inplace=True),
            nn.BatchNorm2d(d),

            nn.Conv2d(d, d*2, 3, 2, 1),
            nn.LeakyReLU(0.1, inplace=True),
            nn.BatchNorm2d(d*2),

            nn.Conv2d(d*2, d*4, 3, 2, 1),
            nn.LeakyReLU(0.1, inplace=True),
            nn.BatchNorm2d(d*4),

            nn.Conv2d(d*4, d*8, 3, 2, 1),
            nn.LeakyReLU(0.1, inplace=True),
            nn.BatchNorm2d(d*8),

            nn.Conv2d(d*8, 1, 3, 2, 1),
        )

    def forward(self, input, label):
        x = torch.cat([input, label], 1)
        x = self.convs(x)
        return x


# Dataset class for noisy and clean images
class ImageDenoisingDataset(Dataset):
    def __init__(self, noisy_dir, clean_dir, noise_levels, transform=None):
        self.noisy_dir = noisy_dir
        self.clean_dir = clean_dir
        self.noisy_images = sorted(os.listdir(noisy_dir))
        self.clean_images = sorted(os.listdir(clean_dir))

        # Ensure the number of noisy and clean images are the same
        min_images = min(len(self.noisy_images), len(self.clean_images))
        self.noisy_images = self.noisy_images[:min_images]
        self.clean_images = self.clean_images[:min_images]

        self.noise_levels = noise_levels  # List of noise levels (e.g., [0.05, 0.1, 0.15])
        self.transform = transform

    def __len__(self):
        return len(self.noisy_images)

    def __getitem__(self, idx):
        noisy_img_path = os.path.join(self.noisy_dir, self.noisy_images[idx])
        clean_img_path = os.path.join(self.clean_dir, self.clean_images[idx])

        noisy_img = Image.open(noisy_img_path).convert('RGB')
        clean_img = Image.open(clean_img_path).convert('RGB')

        # Add random noise based on selected noise level
        noise_level = random.choice(self.noise_levels)
        noisy_img = self.add_noise(noisy_img, noise_level)

        if self.transform:
            noisy_img = self.transform(noisy_img)
            clean_img = self.transform(clean_img)

        return noisy_img, clean_img

    def add_noise(self, img, noise_level):
        """
        Add random Gaussian noise to an image.
        noise_level: the standard deviation of the Gaussian noise
        """
        img = np.array(img) / 255.0  # Normalize to [0, 1]
        noise = np.random.normal(0, noise_level, img.shape)  # Add Gaussian noise
        noisy_img = img + noise
        noisy_img = np.clip(noisy_img, 0, 1)  # Ensure the values stay between [0, 1]
        return Image.fromarray((noisy_img * 255).astype(np.uint8))  # Convert back to uint8
```
